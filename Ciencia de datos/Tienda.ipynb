# Filtros y búsquedas #

#### **🟡 Ejercicio 12: Búsqueda de texto parcial con .str.contains()**  
**Objetivo:** Filtra el DataFrame para obtener solo las filas donde el Cliente contenga la palabra 'Corp' (como TechCorp).

**Pista:** Pista: Usa .str.contains('Corp', na=False).

```python
df_tienda[df_tienda['Cliente'].str.contains('Corp', na=False)]
```

#### **🟡 Ejercicio 13: Filtrado que empieza por... con .str.startswith()**   

**Objetivo:** Filtra el DataFrame para traer las filas donde la Region comience con la letra 'O' (en tus datos esto te devolverá los registros de la región 'Oeste').

**Pista:** La estructura es idéntica al ejercicio anterior, pero cambiando el método por .str.startswith('...').

```python
df_tienda[df_tienda['Region'].str.startswith('Oe')]
```

#### **🟡 Ejercicio 14: Combinando texto y condiciones numéricas**   

**Objetivo:** Encuentra los registros donde el Cliente contenga la palabra 'Labs' (como AlphaLabs) Y que además el Pago_Total sea estrictamente mayor a 1000.

**Pista:** Usa el operador & para unir la condición de texto (.str.contains('Labs', na=False)) y la condición matemática (> 1000). Recuerda envolver cada bloque en sus propios paréntesis ().

```python
df_tienda[(df_tienda['Cliente'].str.contains('Labs', na=False)) & (df_tienda['Pago_Total'] > 1000)]
```

#### **🟡 Ejercicio 15: Filtrado avanzado por exclusión con ~ (Virgulilla)**   

**Objetivo:** Trae todos los registros cuya Categoria NO esté en la lista ['Hardware', 'Software'] (para tus datos, esto debería devolver solo las filas de 'Soporte').

**Pista:** Usa la estructura del Ejercicio 11 con .isin(), pero coloca el símbolo ~ (alt + 126 en Windows / option + ñ en Mac) justo antes de la condición de la columna para invertir el filtro (para decir "que NO esté en").

```python
df_tienda[~df_tienda['Categoria'].isin(['Hardware', 'Software'])]
```

#### **🟡 Ejercicio 16: El filtro de Prioridad Máxima**

**Objetivo:** Muestra los pedidos que tengan Prioridad igual a 'Alta' Y que además hayan aplicado descuento (Descuento_Aplicado igual a 'Si').

**Pista:** Usa el operador & para unir ambas condiciones exactas. Recuerda aislar cada una en sus propios paréntesis ().

```python
df_tienda[(df_tienda['Prioridad'] == 'Alta') & (df_tienda['Descuento_Aplicado'] == 'Si')]
```

#### **🟡 Ejercicio 17: Ordenamiento básico ascendente**.

#### **🟡 Ejercicio 18: Ordenamiento descendente (Los más caros primero)**.  
**Contexto:** Finanzas quiere auditar las transacciones que generaron los mayores ingresos para la tienda.

```python
df_tienda.sort_values(by='Pago_Total', ascending=False)
```

#### **🟡 Ejercicio 19: Ordenamiento jerárquico por múltiples columnas**.  
**Contexto:** Para un reporte de operaciones, queremos organizar los datos por regiones. Pero dentro de cada región, queremos ver los pedidos ordenados desde el precio unitario más alto al más bajo.

**Objetivo:** Ordena el DataFrame por dos columnas: primero por Region de forma alfabética (ascendente), y segundo por Precio_Unitario de mayor a menor (descendente).

**Pista:** Pasa una lista a ambos argumentos: by=['Columna1', 'Columna2'] y ascending=[True, False].

```python
df_tienda.sort_values(by=['Region', 'Precio_Unitario'], ascending=[True, False])
```

#### **🟡 Ejercicio 20: Combinando Filtro y Ordenamiento (Nivel Medio)**.  
**Contexto:** El equipo de ventas quiere analizar el rendimiento en la región Sur, pero quiere enfocarse primero en los pedidos que compraron más volumen.

**Objetivo:** Filtra el DataFrame para obtener solo las filas de la Region 'Sur' Y, sobre ese resultado, ordénalo de mayor a menor según las Unidades.

**Pista:** Puedes encadenar los métodos: df_tienda[condicion].sort_values(...).

```python
df_tienda[df_tienda['Region'] == 'Sur'].sort_values(by=['Unidades'],ascending=False)
```
