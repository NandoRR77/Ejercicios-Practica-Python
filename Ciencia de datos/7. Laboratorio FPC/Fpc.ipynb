### 🛠️ Ejercicio 2: Análisis Disciplinario y Filtrado Condicional Avanzado

**Contexto:**
En el fútbol colombiano, la intensidad del juego físico suele verse reflejada en la cantidad de amonestaciones y expulsiones. Un analista de rendimiento deportivo no solo busca qué equipo anota más goles, sino qué instituciones logran mantener un equilibrio disciplinario o, por el contrario, cuáles suelen jugar al límite del reglamento.

**Enunciado:**
Realiza las siguientes operaciones en tu cuaderno:
1. Crea una nueva columna llamada **`Total_Tarjetas`** que sea la suma de `Tarjetas_Amarillas` y `Tarjetas_Rojas`.
2. Filtra el DataFrame para encontrar únicamente a los equipos que cumplan **ambas** condiciones simultáneamente:
   * Tener más de **2,500** tarjetas totales en estas cuatro décadas.
   * Haber anotado un promedio de goles por partido (**`Promedio_Goles_Por_Partido`**) superior o igual a **1.35**.
3. Guarda este filtro en una variable llamada **`equipos_intensos`** e imprime en pantalla únicamente el nombre del `Equipo`, sus `Total_Tarjetas` y su `Promedio_Goles_Por_Partido`.

**Pistas:**
<details>
<summary>💡 Ver pistas para el Ejercicio 2</summary>

1. **Filtrado con múltiples condiciones:** En Pandas, para evaluar dos condiciones al mismo tiempo (operador lógico Y), debes usar el símbolo ampersand `&`. Cada condición debe ir obligatoriamente envuelta entre paréntesis, siguiendo esta estructura:  
   `df[(condicion_1) & (condicion_2)]`
2. **Operadores de comparación:** Recuerda que "superior o igual" se escribe mediante el operador `>=` y "más de" utilizando el operador `>`.
</details>


```python
#Crear columnas Total_Tarjetas
df_fpc['Total_Tarjetas'] = df_fpc['Tarjetas_Amarillas'] + df_fpc['Tarjetas_Rojas']

#Mostrar nueva columna
df_fpc['Total_Tarjetas']
```

```python
#Filtrar el dataframe con las condiciones dadas y llevar a un nuevo dataframe
df_equipos_intensos = df_fpc[(df_fpc['Total_Tarjetas'] > 2500) & (df_fpc['Promedio_Goles_Por_Partido'] >= 1.35)].copy()

#Imprimir nuevo dataframe solo con las columnas Equipo, Total_Tarjetas y Promedio_Goles_Por_Partido

df_equipos_intensos[['Equipo', 'Total_Tarjetas', 'Promedio_Goles_Por_Partido']] 

```

### 🛠️ Ejercicio 3: Rendimiento por Región Geográfica (Group By)

**Contexto:**
El fútbol en Colombia está fuertemente marcado por el regionalismo y los diferentes estilos de juego de cada zona geográfica (el juego táctico del interior, el despliegue físico del eje cafetero, la velocidad de la costa, etc.). Para entender la geopolítica del FPC, un analista no mira clubes individuales, sino el aporte global de cada región.

**Enunciado:**
Utiliza las funciones de agregación de Pandas para responder a la siguiente pregunta: **¿Cuál es la región que acumula la mayor cantidad de Puntos Acumulados y cuántos Títulos de Liga ha aportado en total?**

Para resolverlo, debes:
1. Agrupar los datos de tu DataFrame original utilizando la columna **`Region`**.
2. Calcular simultáneamente la **Suma Total** de `Puntos_Acumulados` y la **Suma Total** de `Titulos_Ligax40Anos` para cada una de las regiones.
3. Almacenar el resultado en una variable llamada **`rendimiento_regional`** y ordenarlo de mayor a menor según los puntos acumulados.

**Pistas:**
<details>
<summary>💡 Ver pistas para el Ejercicio 3</summary>

1. **Agrupación múltiple:** Para agrupar y aplicar cálculos a columnas específicas al mismo tiempo, la forma más limpia en Pandas es usar `.groupby('Columna_Grupo').agg({'Columna1':'sum', 'Columna2':'sum'}).reset_index()`.
2. **El método `.agg()`:** Este método recibe un diccionario de Python donde la "clave" es la columna que quieres calcular, y el "valor" es la operación matemática en formato de texto (como `'sum'`, `'mean'`, etc.).
3. **Ordenar resultados:** No olvides usar `.sort_values(by='Columna_Criterio', ascending=False)` al final para ordenar de mayor a menor.
</details>


```python
df_rendimiento_regional = df_fpc.groupby('Region')[['Puntos_Acumulados', 'Titulos_Ligax40Anos']].agg('sum').sort_values('Puntos_Acumulados', ascending=False).reset_index().copy()
df_rendimiento_regional
```

### 📊 Ejercicio 5: Gráfico de Dispersión (Scatter Plot) y Correlación

**Contexto:**
En la ciencia de datos deportivos existe un debate eterno: ¿Llenar los estadios y tener una masa gigante de hinchas garantiza ganar campeonatos, o se puede ser un equipo altamente eficiente y ganador con una hinchada más modesta? Para responder esto, los analistas usan gráficos de dispersión que permiten cruzar dos variables numéricas continuas e identificar si existe una relación visual (correlación).

**Enunciado:**
Utilizando tu DataFrame original (**`df_fpc`**), crea un **Gráfico de Dispersión (Scatter Plot)** utilizando `seaborn` que cruce el apoyo en las tribunas con el éxito deportivo. El gráfico debe cumplir los siguientes requisitos:
1. El eje horizontal (X) debe mostrar la variable **`Hinchas_Promedio_Por_Torneo`**.
2. El eje vertical (Y) debe mostrar la variable **`Puntos_Acumulados`**.
3. Para identificar a qué equipo corresponde cada punto del mapa, utiliza el parámetro `hue` para cambiar el color del punto según el **`Equipo`**.
4. Aumenta el tamaño de los puntos usando el parámetro `s=200` para que la visualización sea completamente legible.
5. Agrega un título descriptivo y etiquetas personalizadas a los ejes.

**Pistas:**
<details>
<summary>💡 Ver pistas para el Ejercicio 5</summary>

1. **Función de Dispersión:** En Seaborn, el gráfico de dispersión se genera con la función `sns.scatterplot()`.
2. **Estructura del código:** La sintaxis base que necesitas es:  
   `sns.scatterplot(data=df_fpc, x='Columna_X', y='Columna_Y', hue='Columna_Color', s=200)`
3. **Mover la leyenda:** Como tenemos 10 equipos, la leyenda de colores podría tapar los puntos. Puedes moverla fuera del gráfico usando Matplotlib justo antes del `plt.show()` con la línea:  
   `plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')`





</details>


```python
#Tamaño del gráfico
plt.figure(figsize=(11, 6))

#Seleccionar el tipo de gráfico scatterplot para realizar un diagrama de dispersión
sns.scatterplot(data= df_fpc,
                x='Hinchas_Promedio_Por_Torneo',
                y='Puntos_Acumulados',
                hue='Equipo',
                s=200)

#Titulos y etiquetas
plt.title('¿Influyen las Tribunas en la Tabla? Hinchas vs Puntos', fontsize=14, fontweight='bold')
plt.xlabel('Hinchas Promedio por Torneo', fontweight='bold')
plt.ylabel('Puntos Acumulados (Últimas 4 Décadas)', fontweight='bold')

#Leyenda del gráfico
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', title='Equipos')

#Mostrar el gráfico
plt.show()
```
