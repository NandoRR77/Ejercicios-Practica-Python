## **🍿 Desafío 6: Operaciones matemáticas entre columnas** ##

## **🍿 Desafío 7: Etiquetas condicionales binarias** ##

**Contexto:** El equipo de marketing quiere segmentar a los usuarios según su nivel de consumo para una campaña.

**Objetivo:** Crea una nueva columna llamada Perfil_Consumo. Si el usuario tiene más de 60 Horas_Reproducidas, la columna debe registrar el texto 'Intenso'; de lo contrario, debe registrar 'Moderado'.  

```python
#importar numpy
import numpy as np

#Estructura: df_streaming['Nueva_Columna'] = np.where(df['Columna']condicion_logica, 'Valor_Si_Cumple', 'Valor_Si_No_Cumple')
df_streaming['Perfil_Consumo'] = np.where(df_streaming['Horas_Reproducidas'] > 60, 'Intenso','Moderado')
df_streaming
```

## **🍿 Desafío 8: Transformación y estandarización de texto** ##

**Contexto:** Para evitar errores en las bases de datos de correos masivos, el equipo de sistemas requiere estandarizar el formato de los planes.

**Objetivo:** Modifica la columna existente Tipo_Suscripcion para que todos sus valores queden guardados completamente en letras minúsculas (por ejemplo, transformar 'Premium' en 'premium').

**Pista:** Pandas tiene un accesorio especial para trabajar con textos llamado .str. Para convertir a minúsculas se usa el método .lower(). La estructura para sobrescribir una columna es:df['Columna'] = df['Columna'].str.lower()

```python
df_streaming['Tipo_Suscripcion'] = df_streaming['Tipo_Suscripcion'].str.lower()
df_streaming
```

## **🍿 Desafío 9: Multiplicación condicional (Cálculo de Comisiones/Impuestos)** ##

**Contexto:** El departamento de finanzas va a aplicar un impuesto digital del 10% (0.10) sobre el costo del plan mensual, pero únicamente a los usuarios de la región 'Norte'. A las demás regiones se les aplicará un 0% (0).

**Objetivo:** Crea una nueva columna llamada Impuesto_Digital que calcule el valor en dinero de ese impuesto (es decir, multiplicar Monto_Mensual_USD * 0.10 si es del Norte, o guardar 0 si es de otra región).

```python
df_streaming['Impuesto_Digital'] = np.where(df_streaming['Region_Usuario']=='Norte', df_streaming['Monto_Mensual_USD']*0.10, 0)
df_streaming
```

## **🍿 Desafío 10: Múltiples condiciones avanzadas con np.select()** ##

**Contexto:** El equipo de soporte quiere clasificar las cuentas por su nivel de riesgo de saturación según los dispositivos que conectan simultáneamente.

**Objetivo:** Crea una nueva columna llamada Riesgo_Conexion bajo la siguiente lógica de tres niveles:    

**Si Dispositivos_Conectados es mayor o igual a 5, la etiqueta debe ser 'Alto'.    
Si Dispositivos_Conectados es igual a 3 o 4, la etiqueta debe ser 'Medio'.  
Si es menor o igual a 2, la etiqueta debe ser 'Bajo'.**

**Pista:** La lógica de np.select() funciona creando dos listas independientes: una para tus preguntas lógicas (condiciones) y otra para las respuestas (opciones). Ambas listas deben tener el mismo orden.

```python
# 1. Creas la lista con todas tus condiciones lógicas
condiciones = [
    (df_streaming['Dispositivos_Conectados'] >= 5),  # Condición 1 (Alto)
    ((df_streaming['Dispositivos_Conectados'] == 3) | (df_streaming['Dispositivos_Conectados'] == 4)),  # Condición 2 (Medio)
    (df_streaming['Dispositivos_Conectados'] <= 2)   # Condición 3 (Bajo)
]

# 2. Creas la lista con los resultados correspondientes en el mismo orden
opciones = ['Alto', 'Medio', 'Bajo']

# 3. Aplicas np.select asignándolo a la nueva columna
df_streaming['Riesgo_Conexion'] = np.select(condiciones, opciones, default='Desconocido')

#Mostrar nueva columna
df_streaming
```
