## **🍿 Desafío 10: Múltiples condiciones avanzadas con np.select()** ##

**Contexto:** El equipo de soporte quiere clasificar las cuentas por su nivel de riesgo de saturación según los dispositivos que conectan simultáneamente.

**Objetivo:** Crea una nueva columna llamada Riesgo_Conexion bajo la siguiente lógica de tres niveles:    

**Si Dispositivos_Conectados es mayor o igual a 5, la etiqueta debe ser 'Alto'.    
Si Dispositivos_Conectados es igual a 3 o 4, la etiqueta debe ser 'Medio'.  
Si es menor o igual a 2, la etiqueta debe ser 'Bajo'.**

**Pista:** La lógica de np.select() funciona creando dos listas independientes: una para tus preguntas lógicas (condiciones) y otra para las respuestas (opciones). Ambas listas deben tener el mismo orden.

```python
# 1. Creas la lista con todas tus condiciones lógicas
condiciones = [
    (df_streaming['Dispositivos_Conectados'] >= 5),  # Condición 1 (Alto)
    ((df_streaming['Dispositivos_Conectados'] == 3) | (df_streaming['Dispositivos_Conectados'] == 4)),  # Condición 2 (Medio)
    (df_streaming['Dispositivos_Conectados'] <= 2)   # Condición 3 (Bajo)
]

# 2. Creas la lista con los resultados correspondientes en el mismo orden
opciones = ['Alto', 'Medio', 'Bajo']

# 3. Aplicas np.select asignándolo a la nueva columna
df_streaming['Riesgo_Conexion'] = np.select(condiciones, opciones, default='Desconocido')

#Mostrar nueva columna
df_streaming
```
