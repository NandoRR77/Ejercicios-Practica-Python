```python
import pandas as pd

df_data = pd.read_csv('/Users/nando/Ejercicios-Practica-Python/Ciencia de datos/Money Control/MoneyControl_14.10.2026.csv', sep=";" , on_bad_lines='skip')
df_data.head(10)
```

# Limpieza de datos #

```python
df_data.info()
```

```python
#Contabilizar registros vacíos por columna
df_data.isnull().sum()
```

```python
#Eliminar columnas Persona y Grupo porque están vacías 
df_data = df_data.drop(columns=['Persona', 'Grupo'], errors='ignore')
df_data.head(5)
```

```python
#Convertir Fecha de forma segura (indicando el formato día/mes/año)
df_data['Fecha'] = pd.to_datetime(df_data['Fecha'], format='%d/%m/%y', errors='coerce')
df_data.head(5)
```

```python
#Limpieza del Monto (Especial para formato COP con espacio tras el menos: "- 10.790.850,00")
df_data['Monto'] = df_data['Monto'].astype(str)
df_data['Monto'] = df_data['Monto'].str.replace(' ', '', regex=False)     # Quita espacios (transforma "- " en "-")
df_data['Monto'] = df_data['Monto'].str.replace('.', '', regex=False)     # Quita puntos de miles
df_data['Monto'] = df_data['Monto'].str.replace(',', '.', regex=False)     # Cambia la coma decimal por punto
df_data['Monto'] = pd.to_numeric(df_data['Monto'], errors='coerce')       # Convierte a número flotante

df_data.head(5)
```

```python
#Visualizar registros vacíos para columna categoria y revisar si corresponden todos a Tipo transferencia
df_data[((df_data['Categoría'].isna()) & (df_data['Tipo'] == 'Transferencia'))]
```

```python
#Imputar registros vacíos de la columna Categoria con Sin Clasificar
df_data['Categoría'] = df_data['Categoría'].fillna('Sin Clasificar')
df_data[df_data['Categoría'] == 'Sin Clasificar']
```

### Ajuste de Ruta: Filtro de Gastos Ejecutados ###

Tu tarea en Pandas:
Modifica tu DataFrame de gastos filtrados (el que ya excluye los 'Registro inicial egreso') para que solo conserve las transacciones cuya fecha sea menor o igual al día de hoy.   

• Pistas de Pandas: Puedes importar la librería datetime o usar pd.to_datetime('today') (o la fecha exacta de hoy: 2026-10-14) para hacer la comparación. La estructura sería algo como: df[df['Fecha'] <= '2026-10-14']

```python
#1. Filtrar Tipo Gastos y Categoría diferente a Registro inicial de egreso
#Guardar en nuevo dataframe
df_gastos_filtrados = df_data[(df_data['Tipo'] == 'Gastos') & (df_data['Fecha'] <= '2026-10-31') & ~(df_data['Categoría'] == 'Registro inicial egreso')]
df_gastos_filtrados.head(5)
```

```python
#2. Agrupar y sumar
#3. Ordenar
#4. Aplicar valor absoluto para omitir signo -
df_gastos_filtrados.groupby('Categoría')['Monto'].sum().abs().sort_values(ascending=False)
```

### Excluir categorias que no aplican a los gastos porque son favores

```python
#Excluimos los favores y servicios de mamá que te reembolsan
df_gastos_reales = df_gastos_filtrados[~df_gastos_filtrados['Categoría'].isin(['Favores', 'Servicios Mamá'])]
df_gastos_reales.head(10)
```

### Aplicamos nuevamente el ordenamiento

```python
#2. Agrupar y sumar
#3. Ordenar
#4. Aplicar valor absoluto para omitir signo -
df_gastos_reales.groupby('Categoría')['Monto'].sum().abs().sort_values(ascending=False)
```

# 🛠️ Ejercicio 2: El Impacto de lo "Inclasificable"

### 📌 Contexto Financiero
Cuando atravesamos dificultades financieras, cada peso debe tener nombre y apellido. Dejar transacciones en una categoría genérica como "Sin Clasificar" u "Otros" es el equivalente a dejar la puerta de la casa abierta por la noche: no sabes qué se está saliendo por ahí. Si la suma de estos gastos es alta, significa que tienes una fuga invisible de dinero que no estás controlando.

### 📝 Enunciado
Calcula cuánto dinero total se ha ido en la categoría `"Sin Clasificar"` (que asignamos en la limpieza a los valores nulos). Determina también qué porcentaje representa este monto respecto al gasto total real de tu historial (excluyendo los registros iniciales pero manteniendo tus categorías de gasto).

### 💡 Pistas de Pandas
1. **Para calcular el gasto total real:** Puedes usar la suma de la columna `['Monto']` del DataFrame filtrado que creaste en el Ejercicio 1 (`df_gastos_reales`).
2. **Para filtrar la categoría específica:** Selecciona las filas de `df_gastos_reales` donde la columna `Categoría` sea exactamente igual a `'Sin Clasificar'` y aplica `.sum()` a la columna `['Monto']`.
3. **Para el porcentaje:** Divide el total de "Sin Clasificar" entre el gasto total real y multiplícalo por 100.


```python
#1. Calculo el total de los gastos
total_gastos = df_gastos_reales['Monto'].sum()
total_gastos = total_gastos * -1
print(f'El total de gastos es: {total_gastos:2f}')
```

```python
#2. Filtrar categoria 'Sin clasificar' y sumar el monto
monto_gastos_sin_clasificar = df_gastos_reales[df_gastos_reales['Categoría'] == 'Sin Clasificar']['Monto'].sum()
print(f'El total de gastos sin clasificar es: {monto_gastos_sin_clasificar}')
```

```python
#Calcular el porcentaje de gastos sin clasificar
porcen = monto_gastos_sin_clasificar / total_gastos
print(f'El % de gastos sin clasificar es: {porcen}')
```
