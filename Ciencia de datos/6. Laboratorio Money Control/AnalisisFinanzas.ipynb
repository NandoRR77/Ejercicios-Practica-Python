# 🛠️ Ejercicio 9: Radiografía Evolutiva de Pagos Bancarios (Mes a Mes)

### 📌 Contexto Financiero
Auditar un solo mes nos dice qué falló hoy, pero analizar los pagos bancarios mes a mes nos revela la tendencia del endeudamiento. Si los pagos a los bancos crecen mes a mes, significa que la bola de nieve de las deudas se está acelerando. Necesitamos ver cómo se distribuyeron las salidas de la categoría "Pago bancos y deudas" a lo largo de todo el año 2026.

### 📝 Enunciado
Filtra el DataFrame para aislar todas las transacciones de la categoría `"Pago bancos y deudas"`. Con estos datos, genera una matriz mensual que muestre la suma de los montos en valor absoluto, desglosada por cada mes del año (`Mes`) y clasificada por la `Cuenta` o la `Nota` que generó el desembolso.

### 💡 Pistas de Pandas
<details>

1. **index (Las Filas):** Es la variable que quieres ver de arriba hacia abajo en el extremo izquierdo. En tu caso, queremos ver las cuentas o deudas: `Cuenta`.

2. **columns (Las Columnas):** Es la variable que quieres que se desfile de izquierda a derecha en la parte superior. En tu caso, la línea de tiempo: `Mes`.

3. **values (Los Valores):** Es la columna numérica que se va a repartir dentro de la cuadrícula de la matriz: `Monto`.

4. **aggfunc (La Operación):** Es la función matemática que va a procesar los valores. Como queremos acumular los pesos pagados, usamos la suma: `sum`.


**Estructura Python**:  

DataFrame.pivot_table(index='Fila', columns='Columna', values='Monto_Numérico', aggfunc='Operación')


</details>

```python
#Crear df solo pagos bancos con el df_gastos_reales que es donde he calculado las fechas
df_pagos_bancos = df_gastos_reales[df_gastos_reales['Categoría'] == 'Pago bancos y deudas'].copy()

df_pagos_bancos.head(5)
```

```python
#Crear la tabla pivote
df_pagos_bancos.pivot_table(index='Cuenta', columns='Mes', values='Monto', aggfunc=sum).abs().fillna('-')
```

```python
#Sin pivot_table
df_pagos_bancos.groupby(['Cuenta', 'Mes'])['Monto'].sum().abs()
```
