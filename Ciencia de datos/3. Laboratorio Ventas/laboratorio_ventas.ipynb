```python
import io
import pandas as pd


#crear los csv con código
csv_ventas = """ID_Pedido,Cliente,Plan_Suscripcion,Compras_2025,Total_Gastado_USD,Region
1001, TechCorp ,Premium,12,1200.50,Norte
1002,AlphaLabs,Basico,4,450.00,Sur
1003,BetaOmega,Familiar,20,,Este
1001, TechCorp ,Premium,12,1200.50,Norte
1005,AlphaLabs,Basico,,250.00,Sur
1006,BetaOmega,Familiar,8,600.00,Este
1007,GlobalNet,Basico,2,150.00,
1008,DeltaSystems,Premium,15,1450.00,Oeste
1009,AlphaLabs,Basico,4,450.00,Sur
1010,GlobalNet,Familiar,25,950.00,Oeste"""

csv_envios = """ID_Pedido,Dias_Entrega,Estado_Envio
1001,3,Entregado
1002,5,Entregado
1003,7,Retrasado
1005,4,Entregado
1006,8,Retrasado
1007,2,Entregado
1008,3,Entregado
1010,6,Entregado"""


#Crear los dataframes
df_ventas = pd.read_csv(io.StringIO(csv_ventas))
df_envios = pd.read_csv(io.StringIO(csv_envios))
```

```python
#Mostrar los dataframes
df_ventas
```

```python
#Mostrar los dataframes
df_envios
```

## - [ ] **Fase 1: Diagnóstico y Limpieza de Texto**
  - Detectar celdas vacías (`NaN`) en todo el dataset.

```python
#Detectar celdas vacías (`NaN`) en todo el dataset.
df_ventas.isnull().sum()
```

  - Remover espacios en blanco invisibles (al inicio/final) en la columna `Cliente`.

```python
#Remover espacios en blanco invisibles (al inicio/final) en la columna `Cliente`.

#Buscar los espacios en blanco - Método 1 (El argumento na=False evita errores si tienes valores nulos o NaN en la columna).
df_ventas[df_ventas['Cliente'].str.contains(r"\s", na=False)]
```

```python
#Buscar los espacios en blanco - Método 2
df_ventas[df_ventas['Cliente'].str.isspace() == True]
```

```python
#Remover los espacios en blanco
df_ventas['Cliente'] = df_ventas['Cliente'].str.strip()
df_ventas
```

```python
#Comprobar nuevamente si hay espacios en blanco
df_ventas[df_ventas['Cliente'].str.contains(r"\s", na=False)]
```

## - [ ] **Fase 2: Auditoría de Clientes**
  - Identificar y eliminar filas duplicadas idénticas en `df_ventas`.

```python
#Identificar filas duplicadas identicas
df_ventas[df_ventas.duplicated()]
```

```python
#Conservo los registros en un nuevo dataframe
df_clientes_duplicados = df_ventas[df_ventas.duplicated()]

#Mostrar nuevo dataframe
df_clientes_duplicados
```

```python
#Eliminar los registros duplicados del dataframe original
df_ventas.drop_duplicates(inplace=True)

#Mostrar dataframe depurado
df_ventas
```

## - [ ] **Fase 3: Imputación Estadística**
  - Calcular el promedio de la columna `Compras_2025`.
  - Rellenar los valores vacíos de esa columna con la media calculada.

```python
#Calcular la meedia de compras de 2025
media_compras = df_ventas['Compras_2025'].mean()
print(f'La media de compras de 2025 es: {media_compras}')
```

```python
#Asignar media a valores vacíos en columna Compras_2025
df_ventas['Compras_2025'] = df_ventas['Compras_2025'].fillna(media_compras)

df_ventas
```

## - [ ] **Fase 4: Consolidación Multi-tabla**
  - Unir de forma horizontal (`merge`) la tabla de ventas con la de envíos.
  - Asegurar que no se pierda ningún pedido de la tabla principal (Left Join).

```python
df_unificada = df_ventas.merge(df_envios, on= 'ID_Pedido', how='left')
df_unificada
```

## - [ ] **Fase 5: Reporte de Negocio**
  - Crear una tabla resumen agrupada por `Plan_Suscripcion`.
  - Mostrar la suma del `Total_Gastado_USD` y el promedio de `Compras_2025`.

```python
#.reset_index() quita la columna de agrupación como indice y la vuelve normal
df_unificada.groupby('Plan_Suscripcion')[['Total_Gastado_USD','Compras_2025']].agg({'Total_Gastado_USD': 'sum', 'Compras_2025': 'mean'}).reset_index()
```

```python
#Exportar reporte como csv
'''
index=False: Es un argumento vital. Si no lo pones, Pandas guardará los números de las filas (0, 1, 2...) 
como una columna sin nombre en el archivo final, lo cual ensucia el documento.
'''
df_unificada.to_csv('reporte_retencion_clientes.csv', index=False)
```
